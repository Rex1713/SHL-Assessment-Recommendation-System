```python
import pandas as pd
import re

# Load your cleaned CSV


# Function to convert time strings to minutes
def convert_time_to_minutes(time_str):
    if pd.isnull(time_str):
        return None
    time_str = str(time_str).lower()

    if "Untimed" in time_str:
        return 9999  # Represents untimed assessments

    if "not found" in time_str:
        return None  # Missing data to be handled separately

    match = re.search(r"(\d+)\s*(minute|min)", time_str)
    if match:
        return int(match.group(1))

    match = re.search(r"(\d+)\s*hour", time_str)
    if match:
        return int(match.group(1)) * 60

    return None  # fallback

# Apply the function
df["Duration (Minutes)"] = df["Assessment Length"].apply(convert_time_to_minutes)

# Optional: Fill missing durations ("not found") using average per Type or just global average


# Save cleaned version
df.to_csv("shl_product_catalog_final.csv", index=False)
print("✅ Cleaned data saved to 'shl_product_catalog_final.csv'")
```

```python
import pandas as pd
df = pd.read_csv('./final.csv')
```

```python
# Mapping abbreviation to full name
type_mapping = {
    "A": "Ability & Aptitude",
    "B": "Biodata & Situational Judgement",
    "C": "Competencies",
    "D": "Development & 360",
    "E": "Assessment Exercises",
    "K": "Knowledge & Skills",
    "P": "Personality & Behavior",
    "S": "Simulations"
}
# Function to expand each abbreviation in a comma-separated string
def expand_types_abbreviations(types_str):
    return ", ".join([type_mapping.get(abbr.strip(), abbr.strip()) for abbr in types_str.split(",")])

# Apply to the column
df["Types"] = df["Types"].apply(expand_types_abbreviations)


```
